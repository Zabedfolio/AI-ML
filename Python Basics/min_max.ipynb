### C. Even, Odd, Positive and Negative

```python
n = int(input())

inp = input()

numbers = inp.split()
pos=0; neg=0; even=0; odd=0

for i in range(n):
    x = int(numbers[i])

    if x>0:
        pos+=1
    elif x<0:
        neg+=1

    if x%2==0:
        even+=1
    else:
        odd+=1


print("Even:",even)
print("Odd:",odd)
print("Positive:",pos)
print("Negative:",neg)

```

### Q. Digits

```python
t = int(input())

for i in range(t):
    n = int(input())

    if n == 0:
        print("0")
        continue

    while n>0:
        print(n%10, end=" ")
        n//=10
    print()

```
