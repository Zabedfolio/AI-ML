```python
s = {1, 2, 3, 4, 5}

s.add(10)
s.add(0)
# s.pop()
s.remove(5)

print(s) # {1, 2, 3, 4, 5, 10}
```

### set methods

```python
s = {1, 2, 3, 4, 5}
a = {6, 7, 8}

print(s.union(a)) # {1, 2, 3, 4, 5, 6, 7, 8}

print(s.intersection(a)) # {4, 5}

print(s.isdisjoint(a)) # False
```
