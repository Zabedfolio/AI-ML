#### Tuple -> Collection of data,immutable

```python
# declaration
int_tup = (10., 20, 30, 40, 50)
float_tup = (10.5, 20.5, 30.5, 40.5, 50.5)
mix_tup = (10, 20.5, "Hello", True, None)

print(mix_tup)
```

```python
#access

tup = (10, 20, 30, 40, 50)

# print(tup[0]) # 10

new_tup = tup[0:3]
print(new_tup) # (10, 20, 30)
```

```python
#immutable

tup = (10, 10, 20, 30, 40, 50)

#can't change after a tuiple is created

#tuple methods

print(tup.count(10)) 
print(tup.index(20))
```
