# String Methods

```python
string = "welcome to phitron ML course ml"

processed_string = string.lower()

print(processed_string)

print("phitron" in processed_string)

shuru_index = processed_string.find("ml")

print(shuru_index)

sesher_index = processed_string.rfind("ml")
print(sesher_index)


count = processed_string.count("ml")
print(count)

new_string = string.replace("ML", "AI/ML")

print(new_string)
```
